```python
# 2. Import the library
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
# 3. Load the pre-trained YOLOv8 model (small version)
model = YOLO('yolov8n.pt') # we can use yolov8s.pt, yolov8m.pt, etc.
# 4. Load your image
image_path = 'boubike1.jpg'
image = cv2.imread(image_path)
# Convert image from BGR to RGB (for correct color display with matplotlib)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
# 5. Perform object detection
results = model(image_path)
# 6. Visualize detection results
results[0].plot() # This will draw the bounding boxes on the image
# 7. Show the image
plt.figure(figsize=(12, 8))
plt.imshow(results[0].plot()[..., ::-1]) # Con'6'6vert BGR to RGB for plt
plt.axis('off')
plt.title("Object Detection using YOLOv8")
plt.show()
```

```python
from ultralytics import YOLO
import cv2
from moviepy.editor import VideoFileClip, ImageSequenceClip
from IPython.display import Image, display

# Load YOLO model
model = YOLO("yolov8n.pt")

# Load your video
video_path = "video.mp4"   # put your video in same folder
clip = VideoFileClip(video_path).subclip(0, 5)

frames = []

# Process frames
for frame in clip.iter_frames():
    results = model(frame)
    annotated = results[0].plot()
    frames.append(annotated)

# Save output
output_clip = ImageSequenceClip(frames, fps=clip.fps)
output_clip.write_videofile("output.mp4", audio=False)
output_clip.write_gif("output.gif")

# Display result
with open("output.gif",'rb') as f:
    display(Image(f.read()))
```
